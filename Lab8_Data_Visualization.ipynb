## 8. Filter the data
We create a few filtered subsets to visualize:
- a **quality category** column (Low ≤ 5, Medium = 6, High ≥ 7),
- **high-quality wines** only (quality ≥ 7),
- **high-alcohol wines** (alcohol above the median).

```python
def quality_label(q):
    if q <= 5:
        return "Low"
    elif q == 6:
        return "Medium"
    else:
        return "High"

df["quality_category"] = df["quality"].apply(quality_label)

high_quality = df[df["quality"] >= 7]
high_alcohol = df[df["alcohol"] > df["alcohol"].median()]

print("High-quality wines:", len(high_quality))
print("High-alcohol wines:", len(high_alcohol))
print("Average alcohol – all wines: %.2f%%, high-quality wines: %.2f%%"
      % (df["alcohol"].mean(), high_quality["alcohol"].mean()))
```

## 9. Visualizations with Matplotlib

### 9.1 Histogram – distribution of alcohol content

```python
plt.hist(df["alcohol"], bins=20, color="steelblue", edgecolor="black")
plt.title("Distribution of Alcohol Content in Red Wines")
plt.xlabel("Alcohol (% vol)")
plt.ylabel("Number of wines")
plt.show()
```

**Observation:** Most wines have between about 9% and 11% alcohol, and the distribution is right-skewed (fewer wines with very high alcohol).

### 9.2 Bar chart – number of wines per quality score

```python
counts = df["quality"].value_counts().sort_index()
plt.bar(counts.index, counts.values, color="darkred")
plt.title("Number of Wines per Quality Score")
plt.xlabel("Quality score")
plt.ylabel("Count")
plt.xticks(counts.index)
plt.show()
```

**Observation:** Most wines are rated 5 or 6; very few are rated 3, 4 or 8, so the dataset is imbalanced.

### 9.3 Scatter plot – alcohol vs. volatile acidity (coloured by quality)

```python
sc = plt.scatter(df["alcohol"], df["volatile acidity"],
                 c=df["quality"], cmap="viridis", alpha=0.6)
plt.colorbar(sc, label="Quality")
plt.title("Alcohol vs. Volatile Acidity")
plt.xlabel("Alcohol (% vol)")
plt.ylabel("Volatile acidity (g/L)")
plt.show()
```

**Observation:** Higher-quality wines (yellow/green points) tend to sit at higher alcohol and lower volatile acidity.

### 9.4 Pie chart – share of each quality category

```python
cat_counts = df["quality_category"].value_counts()[["Low", "Medium", "High"]]
plt.pie(cat_counts, labels=cat_counts.index, autopct="%1.1f%%",
        colors=["#e07a5f", "#f2cc8f", "#81b29a"], startangle=90)
plt.title("Wines by Quality Category")
plt.show()
```

### 9.5 Line plot – average alcohol by quality score

```python
avg_alcohol = df.groupby("quality")["alcohol"].mean()
plt.plot(avg_alcohol.index, avg_alcohol.values, marker="o", color="purple")
plt.title("Average Alcohol Content by Quality Score")
plt.xlabel("Quality score")
plt.ylabel("Average alcohol (% vol)")
plt.show()
```

**Observation:** Average alcohol rises steadily with quality from score 5 upward.

## 10. Visualizations with Seaborn

### 10.1 Count plot – quality categories

```python
sns.countplot(data=df, x="quality_category", order=["Low", "Medium", "High"],
              hue="quality_category", palette="Set2", legend=False)
plt.title("Count of Wines by Quality Category")
plt.xlabel("Quality category")
plt.ylabel("Count")
plt.show()
```

### 10.2 Box plot – alcohol by quality score

```python
sns.boxplot(data=df, x="quality", y="alcohol", hue="quality", palette="coolwarm", legend=False)
plt.title("Alcohol Content by Quality Score")
plt.xlabel("Quality score")
plt.ylabel("Alcohol (% vol)")
plt.show()
```

**Observation:** The median alcohol level is clearly higher for wines rated 7 and 8. Outliers appear mainly in the score-5 group.

### 10.3 Violin plot – volatile acidity by quality category

```python
sns.violinplot(data=df, x="quality_category", y="volatile acidity",
               order=["Low", "Medium", "High"], hue="quality_category",
               palette="muted", legend=False)
plt.title("Volatile Acidity by Quality Category")
plt.xlabel("Quality category")
plt.ylabel("Volatile acidity (g/L)")
plt.show()
```

**Observation:** Low-quality wines have higher and more spread-out volatile acidity (which gives a vinegar-like taste).

### 10.4 Heatmap – correlation between features

```python
plt.figure(figsize=(11, 8))
corr = df.drop(columns=["quality_category"]).corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0, linewidths=0.5)
plt.title("Correlation Matrix of Wine Features")
plt.show()
```

```python
# Features most correlated with quality
corr["quality"].drop("quality").sort_values(ascending=False).round(2)
```

**Observation:** Alcohol has the strongest positive correlation with quality, and volatile acidity has the strongest negative correlation.

### 10.5 Regression plot – alcohol vs. quality

```python
sns.regplot(data=df, x="alcohol", y="quality",
            scatter_kws={"alpha": 0.3}, line_kws={"color": "red"})
plt.title("Alcohol vs. Quality with Trend Line")
plt.xlabel("Alcohol (% vol)")
plt.ylabel("Quality score")
plt.show()
```

### 10.6 Histogram with KDE – filtered data: high-quality vs. all wines

```python
sns.histplot(df["sulphates"], color="grey", label="All wines", kde=True, stat="density")
sns.histplot(high_quality["sulphates"], color="green", label="High quality (≥7)",
             kde=True, stat="density", alpha=0.5)
plt.title("Sulphates: All Wines vs. High-Quality Wines")
plt.xlabel("Sulphates (g/L)")
plt.legend()
plt.show()
```

### 10.7 Pair plot – key features by quality category

```python
sns.pairplot(df[["alcohol", "volatile acidity", "sulphates", "citric acid", "quality_category"]],
             hue="quality_category", hue_order=["Low", "Medium", "High"],
             palette="Set1", plot_kws={"alpha": 0.5})
plt.suptitle("Pair Plot of Key Features", y=1.02)
plt.show()
```

### 10.8 Scatter plot of filtered data – high-alcohol wines only

```python
sns.scatterplot(data=high_alcohol, x="citric acid", y="volatile acidity",
                hue="quality_category", hue_order=["Low", "Medium", "High"],
                palette="Set1", alpha=0.7)
plt.title("High-Alcohol Wines: Citric Acid vs. Volatile Acidity")
plt.xlabel("Citric acid (g/L)")
plt.ylabel("Volatile acidity (g/L)")
plt.show()
```

Key findings: most wines are rated 5 or 6; alcohol content is the feature most positively associated with quality; volatile acidity is the most negatively associated; and high-quality wines tend to have slightly higher sulphates.
